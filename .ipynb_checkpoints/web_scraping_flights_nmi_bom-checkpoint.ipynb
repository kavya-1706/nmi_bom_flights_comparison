```python
len(links)  # 16 destinations, to and fro = 32
```

```python
driver = webdriver.Chrome()
e = []
for link in links:
    driver.get(link)
    time.sleep(5)  
    elements = driver.find_elements(By.CLASS_NAME, "pIav2d")
    for element in elements:
        e.append(element.text)

driver.quit()
```

#### Raw data received from the website

```python
e
```

```python
len(e)
```

```python
with open("nmi_flight_data.txt", "w", encoding="utf-8") as f:
    for flight in e:
        f.write(flight)
        f.write("\n" + "="*50 + "\n")
```

# Data cleaning and preprocessing

```python
flight_data=[]
for flight in e:
    flight = flight.replace('\n', '  ')
    flight_d = flight.split('  ')
    flight_data.append(flight_d)
flight_data = [flight for flight in flight_data if flight != ['']]
```

```python
flight_data
```

```python
flights_with_stops = []
flights_nonstop = []
for flight in flight_data:
  
    if flight[6] == 'Nonstop':
        
        flights_nonstop.append(flight)
    else:
      
        flights_with_stops.append(flight)

print("Flights with stops:")
print(flights_with_stops)
print("\nFlights Nonstop:")
print(flights_nonstop)
```

```python
df1 = pd.DataFrame(flights_with_stops, columns=['Departure_Time', 'Dummy', 'Arrival_Time', 'Airline', 'Duration', 'Route', 'Stops', 'Layover_Time_Location', 'CO2_Emissions', 'Emissions_Change', 'Price'])
df1.drop(columns=['Dummy'], inplace=True)
df1
```

```python
for row in flights_nonstop:
    if len(row) == 11:
        row.pop()

df2 = pd.DataFrame(flights_nonstop, columns=['Departure_Time', 'Dummy', 'Arrival_Time', 'Airline', 'Duration', 'Route', 'Stops', 'CO2_Emissions', 'Emissions_Change', 'Price'])

df2.drop(columns=['Dummy'], inplace=True)

df2['Layover_Time_Location'] = '0 minutes'

df2 = df2[['Departure_Time', 'Arrival_Time', 'Airline', 'Duration', 'Route', 'Stops', 'Layover_Time_Location', 'CO2_Emissions', 'Emissions_Change', 'Price']]

df2
```

```python
df2.drop(3, inplace=True)
df2.reset_index(drop=True, inplace=True)
df2
```

```python
df = pd.concat([df1, df2], ignore_index=True)
df
```

```python
duplicates = df[df.duplicated(keep=False)]
unique_duplicates = duplicates.drop_duplicates()
unique_duplicates
```

```python
df = df.drop_duplicates(keep='first')
df
```

```python
duplicates = df[df.duplicated(keep=False)]
unique_duplicates = duplicates.drop_duplicates()
print(unique_duplicates)
```

Added column 'Class' and value 'economy' since all this data from this webpage consists of economy class flights. 

```python
df['Class'] = 'economy'
df
```

```python
df = df[df["Price"] != "Price unavailable"]
df
```

# Final dataset- saved to csv file

```python
df.to_csv("google_flight_nmi.csv")
```
